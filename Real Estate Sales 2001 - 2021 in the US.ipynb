```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/Users/anpabelt/Data Analytics/Dataset/Real_Estate_Sales_2001-2021_GL 2.csv')
data
```

```python
data.columns
```

```python
data.info()
```

```python
data.drop(columns=['Non Use Code', 'Assessor Remarks', 'OPM remarks', 'Location'], inplace=True)
```

```python
data
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
data['Property Type'] = data['Property Type'].fillna('Unknown')
data['Residential Type'] = data['Residential Type'].fillna('Unknown')
data['Address'] = data['Address'].fillna('Unknown')
data['Date Recorded'] = data['Date Recorded'].fillna('Unknown')
```

```python
data.isna().sum()
```

```python
data
```

```python
data['Assessed Value'] = data['Assessed Value'].astype(int)
data['Sale Amount'] = data['Sale Amount'].astype(int)
```

```python
data
```

```python
data['Date Recorded'] = pd.to_datetime(data['Date Recorded'], format="%m/%d/%Y", errors='coerce')
data['Month'] = data['Date Recorded'].dt.month

data
```

```python
data['Month'] = data['Month'].astype('Int64')

data
```

```python
sales_month = data.groupby('Month').agg({ "Sale Amount": "sum",
}).reset_index()

# Plotting the line chart
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(sales_month['Month'], sales_month['Sale Amount'], marker='o')  # Create line chart
plt.title('Monthly Sum of Sale Amounts')  # Chart title
plt.xlabel('Month')  # X-axis label
plt.ylabel('Sum of Sale Amount')  # Y-axis label
plt.grid(True)  # Add gridlines  # Adjust the layout
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Okt', 'Nov', 'Dec'])
plt.show()
```

```python
sales_year = data.groupby('List Year').agg({ "Sale Amount": "sum",
}).reset_index()

# Plotting the line chart
plt.figure(figsize=(15, 10))  # Set the figure size
plt.plot(sales_year['List Year'], sales_year['Sale Amount'], marker='o')  # Create line chart
plt.title('Yearly Sum of Sale Amounts')  # Chart title
plt.xlabel('Year')  # X-axis label
plt.ylabel('Sum of Sale Amount')  # Y-axis label
plt.grid(True)  # Add gridlines  # Adjust the layout
plt.xticks(range(2001, 2022))
plt.show()
```

```python
data_groupby_town = data.groupby(["Town"]).agg({
    "Assessed Value": "sum", 
    "Sale Amount": "sum",
    "Sales Ratio": "mean"
}).reset_index()
```

```python
data_groupby_town
```

```python
data_groupby_town = data_groupby_town.sort_values(by="Sale Amount", ascending=False)

plt.figure(figsize=(12, 8))

plt.title("Top 10 Town with the Most Sum Sales")

# Create a color palette with a gradient effect
palette = sns.color_palette("Blues_r", n_colors=10)

plt.xticks(rotation=90)


sns.barplot(data=data_groupby_town.head(10), x="Town", y="Sale Amount", palette=palette)

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

```python
data_groupby_town = data_groupby_town.sort_values(by="Assessed Value", ascending=False)

plt.figure(figsize=(12, 8))

plt.title("Top 10 Town with the Most Sum Tax")

palette = sns.color_palette("Blues_r", n_colors=10)

plt.xticks(rotation=90)


sns.barplot(data=data_groupby_town.head(10), x="Town", y="Assessed Value", palette=palette)

plt.xticks(rotation=90)

plt.tight_layout()

plt.show()
```

```python
prop_type_variety = data['Property Type'].value_counts()
prop_type_counts = prop_type_variety.reset_index()
prop_type_counts.columns = ['Property Type', 'Count']

plt.figure(figsize=(12, 8))
sns.barplot(x='Property Type', y='Count', data=prop_type_counts)

# Add title and labels
plt.title('Distribution of Property Types')
plt.xlabel('Property Type')
plt.ylabel('Count')

# Rotate the x-axis labels for better readability
plt.xticks(rotation=90)

# Show the plot
plt.tight_layout()
plt.show()
```

```python
data_corr = data[['List Year', 'Assessed Value', 'Sale Amount']]
data_corr.corr()
```

```python
f, ax = plt.subplots(figsize=(9, 9))
ax.set_title('Year, Sales, and Tax Correlation')
sns.heatmap(data_corr.corr(), annot=True, robust=True, linewidths=.2, fmt= '.2f', ax = ax )
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(x='Assessed Value', y='Sale Amount', hue='Property Type', data=data, alpha=0.5)
plt.title('Sale Amount vs. Assessed Value by Property Type')
plt.xlabel('Assessed Value ($)')
plt.ylabel('Sale Amount ($)')

plt.xlim(data['Assessed Value'].min() * 0.9, data['Assessed Value'].max() * 1.1)
plt.ylim(data['Sale Amount'].min() * 0.9, data['Sale Amount'].max() * 1.1)

plt.legend(title='Property Type')
plt.show()
```
